# Univariate Analysis

Question 1: What does the distribution of customer ratings looks like? Is it skewed?

```python
sns.distplot(df['Rating'])
plt.axvline(x=np.mean(df['Rating']),c='red',ls="--",label='mean')
plt.axvline(x=np.percentile(df['Rating'],25),c='green',ls="--",label='1st Quartile (25%)')
plt.axvline(x=np.percentile(df['Rating'],75),c='black',ls="--",label='3rd Quartile (75%)')
plt.legend()
```

According to the Graph , we can tell that the data is uniform ,i.e  it is not skewed .

Question 4: Is there a noticeable time trend in gross income?

We need to aggregate the data before plotting the line chart , because ther may be many customer in a single date 

```python

# Aggregate the data by date to calculate the mean gross income per day
grouped_data = df['gross income'].resample('D').mean()

# Plot the line chart
plt.figure(figsize=(12, 6))
sns.lineplot(x=grouped_data.index, y=grouped_data.values)
plt.title('Time Trend of Gross Income', fontsize=16)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Gross Income', fontsize=12)
plt.grid(True)
plt.show()
```
